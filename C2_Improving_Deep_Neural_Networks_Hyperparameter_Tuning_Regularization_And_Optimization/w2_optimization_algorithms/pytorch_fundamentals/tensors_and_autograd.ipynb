```python
import numpy as np

# Import torch and print its version.
import torch

print(torch.__version__)
```

```python
# Build a device variable the professional way: use mps if available, else fall back to cpu. (One clean expression or a tiny helper — your choice. The fallback is the professional part; never hard-code mps.)
device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")
```

```python
# Create one tensor from a Python list and one with torch.randn of shape (3, 4).
# For each, print its dtype, shape, and device.
tensor1 = torch.tensor(
    [[1, 2, 3, 4], [3, 4, 5, 6], [0, 1, 2, 3]], dtype=torch.float32, device=device
)
list1 = [[1, 2, 3, 4], [3, 4, 5, 6], [0, 1, 2, 13]]
tensor2 = torch.tensor(list1, dtype=torch.float32)
tensor3 = torch.randn(3, 4)
for tensor in [tensor1, tensor2, tensor3]:
    print(f"\n\nPrinting tensor details for {tensor}:")
    print(f"  dtype: {tensor.dtype}")
    print(f"  shape: {tensor.shape}")
    print(f"  device: {tensor.device}")
```

```python
# Make a NumPy array, bridge it into a tensor, and print that tensor's dtype. Then notice what you got versus the randn tensor's dtype.

np_array = np.random.randn(3, 4)  # floats → NumPy makes these float64
tensor_from_np = torch.from_numpy(np_array)  # the memory-sharing bridge
print(f"Tensor from NumPy array - dtype: {tensor_from_np.dtype}")
print(f"Random tensor dtype: {tensor3.dtype}")
tensor_from_np2 = torch.from_numpy(np_array).float()  # convert to float32
print(
    f"Tensor from NumPy array after conversion to float32 - dtype: {tensor_from_np2.dtype}"
)
```

```python
# Move your randn tensor onto device and confirm it landed there.
tensor3 = tensor3.to(device)
print(f"Random tensor device after moving to device: {tensor3.device}")
```
