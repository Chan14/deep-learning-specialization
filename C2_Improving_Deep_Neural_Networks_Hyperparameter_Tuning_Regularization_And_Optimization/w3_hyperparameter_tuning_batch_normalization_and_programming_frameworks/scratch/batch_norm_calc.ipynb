```python
import numpy as np
```

```python
z_k = np.array([2, 8, 4, 6])
mu = np.mean(z_k)
sigma = np.std(z_k)
print("Mean:", mu)
print("Standard Deviation:", sigma)
```

```python
z_norm = (z_k - mu) / sigma
print("Normalized z_k:", z_norm)
```

```python
gamma = 2
beta = 3
z_tilde = gamma * z_norm + beta
print("Transformed z_tilde:", z_tilde)
```

```python
z_reverted = sigma * z_norm + mu
print("Reverted z_k:", z_reverted)
```

```python
X = np.array([[1, 6, 2], [3, 1, 8], [5, 4, 4]], dtype=float)
W = np.array([[1, 5, 2], [1, 2, 3], [-1, -2, 1], [7, 7, 7]], dtype=float)
b = np.array([1, 2, 4, 5], dtype=float)
```

```python
Z = np.dot(W, X) + b.reshape(-1, 1)
print("Z:", Z)
```

```python
mu = np.mean(Z, axis=1, keepdims=True)
print("mu:", mu)
sigma = np.std(Z, axis=1, keepdims=True)
print("sigma:", sigma)
```

```python
Z_norm = Z - mu
Z_norm = Z_norm / sigma
print("Normalized Z:", Z_norm)
```

```python
W2 = W * 25
print("W2:", W2)
```

```python
Z2 = np.dot(W2, X) + b.reshape(-1, 1)
print("Z2:", Z2)
```

```python
mu2 = np.mean(Z2, axis=1, keepdims=True)
print("mu2:", mu2)
sigma2 = np.std(Z2, axis=1, keepdims=True)
print("sigma2:", sigma2)
Z2_norm = (Z2 - mu2) / sigma2
print("Normalized Z2:", Z2_norm)

print(np.abs(Z2_norm - Z_norm).max())
```
