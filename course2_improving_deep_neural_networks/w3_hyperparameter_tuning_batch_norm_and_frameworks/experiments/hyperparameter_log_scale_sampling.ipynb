```python
import numpy as np

np.set_printoptions(precision=5, suppress=True)
```

```python
low = np.log10(0.0001)  # = -4
high = np.log10(1.0)  # =  0
learning_rates = 10 ** np.linspace(-4, 0, 25)
print(learning_rates)
```

```python
low = np.log10(0.0001)  # = -4
high = np.log10(1.0)  # =  0

exponents = np.random.uniform(low, high, size=25)
print(exponents)
learning_rates = 10**exponents
print(learning_rates)
```

```python
learning_rates = np.random.uniform(0.0001, 1.0, size=12)
print(learning_rates)
```

```python
low = np.log10(0.0001)  # = -4
high = np.log10(1.0)  # =  0

exponents = np.random.uniform(low, high, size=12)
print(exponents)
learning_rates = 10**exponents
print(learning_rates)
print(np.sort(learning_rates))
```

```python
# for momentum, we can sample from a uniform distribution between 0.9 and 0.99
momentums = np.random.uniform(0.9, 0.999, size=12)
print(np.sort(momentums))
```

```python
low = np.log10(0.001)  # 1 - beta ranges from 0.001 to 0.1
high = np.log10(0.1)

r = np.random.uniform(low, high, size=12)
momentums = 1 - 10**r
print(np.sort(momentums))
```

```python
beta_low, beta_high = 0.9, 0.999
low = np.log10(1 - beta_high)
high = np.log10(1 - beta_low)

r = np.random.uniform(low, high, size=100)
betas = 1 - 10**r
print(np.sort(betas))
```

```python
low = np.log10(0.001)  # = -4
high = np.log10(0.1)  # =  0
betas = 1 - (10 ** np.linspace(low, high, 100))
print(betas)
count_range_1 = np.sum((betas >= 0.9) & (betas < 0.99))
count_range_2 = np.sum((betas >= 0.99) & (betas < 0.999))
print(f"Values in [0.9, 0.99):   {count_range_1}")
print(f"Values in [0.99, 0.999]: {count_range_2}")
```
