```python
w = tf.Variable(0, dtype=tf.float32)
optimizer = tf.keras.optimizers.Adam(0.1)


def train_step():
    with tf.GradientTape() as tape:
        cost = w**2 - 10 * w + 25
    trainable_variables = [w]
    grads = tape.gradient(cost, trainable_variables)
    optimizer.apply_gradients(zip(grads, trainable_variables))


print("Before training: w =", w)
```

```python
train_step()
print("After 1 training step: w =", w)
```

```python
for i in range(100):
    train_step()
print("After 100 training steps: w =", w)
```

```python
w = tf.Variable(0.0, dtype=tf.float32)
x = np.array([1.0, -10.0, 25.0], dtype=np.float32)
optimizer = tf.keras.optimizers.Adam(0.1)

print("Before training: w =", w.numpy())


def training(x, w, optimizer):
    with tf.GradientTape() as tape:
        cost = x[0] * w**2 + x[1] * w + x[2]  # forward pass, recorded
    grads = tape.gradient(cost, [w])  # compute dcost/dw
    optimizer.apply_gradients(zip(grads, [w]))  # one update step
```

```python
training(x, w, optimizer)
print("After 1 training step: w =", w.numpy())
```

```python
for i in range(100):
    training(x, w, optimizer)
print("After 100 training steps: w =", w.numpy())
```
